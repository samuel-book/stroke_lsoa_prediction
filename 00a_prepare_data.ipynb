# Prepare admissions and population data

Clean the existing collated MSOA-level admissions and age band population data.

__Inputs:__
+ collated demographics data (including HES admissions numbers and limited age band proportions) by MSOA in England and Wales.
+ collated demographics data (including more age band proportions) by MSOA in England and Wales.
+ MSOA names and codes lookup
 
__Results:__
+ Data: admissions numbers and numbers of people in each age band by MSOA in England.
  + File name: `msoa_cleaned.csv`
  + One row per MSOA,
  + columns for admission numbers, numbers of people by age band etc.

__Method:__
+ Removing Welsh MSOAs, which seem to have invalid HES data.
+ Recalculate the total numbers of patients using the numbers of patients in each health group (bad, fair, good health).
+ Calculate the numbers of patients in each age band by multiplying the new total numbers of patients by the proportions of people in each age band.
+ Use the IMD rankings to group the MSOAs by quantile (e.g. one group with the 20% most deprived, another with the 20-40% most deprived...).
+ Save a copy of this cleaned data.

## Code setup

```python
import os
import polars as pl
import numpy as np
import matplotlib.pyplot as plt
```

## Load data

Patient demographics by MSOA:

```python
path_to_msoa_stats = os.path.join('data', 'msoa_collated.csv')

df_stats_old = pl.read_csv(path_to_msoa_stats)
```

Link this to MSOA names and codes.

Load the data:

```python
path_to_msoa = os.path.join('data', 'Output_Area_to_LSOA_to_MSOA_to_Local_Authority_District_(December_2017)_Lookup_with_Area_Classifications_in_Great_Britain.csv')

df_msoa_names_codes = pl.read_csv(path_to_msoa)

# Remove repeats from all the LSOA and OA:
df_msoa_names_codes = df_msoa_names_codes.unique(subset='MSOA11CD')[['MSOA11CD', 'MSOA11NM']]
```

```python
df_msoa_names_codes.head()
```

```python
df_stats_old = df_stats_old.join(df_msoa_names_codes, left_on='MSOA', right_on='MSOA11NM', how='left')

df_stats_old.head(3)
```

Load newer stats dataframe with more age band info:

```python
path_to_msoa_stats = os.path.join('data', 'collated_data_regional_msoa11cd.csv')

df_stats = pl.read_csv(path_to_msoa_stats)
```

```python
df_stats.head()
```

Only keep columns that match other MSOA dataframe:

```python
cols_to_keep = ['msoa11cd'] + [f'age_{i:.0f}_proportion' for i in range(0, 95, 5)]

df_stats = df_stats[cols_to_keep]
```

Merge together:

```python
df_stats = df_stats_old.join(df_stats, left_on='MSOA11CD', right_on='msoa11cd', how='left')
```

```python
df_stats.head()
```

## Remove Wales

Check sum of admissions for Welsh areas:

```python
df_stats.filter(df_stats['country'] == 'W')['admissions'].sum()
```

Welsh data is always zero so remove it.

```python
df_stats = df_stats.filter(df_stats['country'] != 'W')
```

## Total patients

Recalculate total numbers of patients:

```python
df_stats = df_stats.with_columns((pl.col('good_health') + pl.col('fair health') + pl.col('bad health')).alias('total_health'))
```

```python
df_stats.head()
```

Compare with "all persons" value:

```python
zmax = max([df_stats['All persons'].max(), df_stats['total_health'].max()])
zmin = min([df_stats['All persons'].min(), df_stats['total_health'].min()])

fig, ax = plt.subplots()

# Equality line:
ax.plot([zmin, zmax], [zmin, zmax], color='k', linewidth=1)
# Data:
ax.scatter(df_stats['All persons'], df_stats['total_health'], s=2, alpha=0.3)

ax.set_xlabel('All persons')
ax.set_ylabel('total_health')
ax.set_aspect('equal')

plt.show()
```

Most data is below the equality line so the numbers of people in the "total health" column is usually less than the numbers in the "all persons" column.

## Combine age bands

Merge age bands under 65:

```python
ages_to_merge = [f'age_{i:.0f}_proportion' for i in range(0, 65, 5)]

df_stats = df_stats.insert_column(-1, pl.Series('age_less65_proportion', df_stats[ages_to_merge].sum_horizontal()))

df_stats = df_stats.drop(ages_to_merge)
```

Merge age bands over 80:

```python
ages_to_merge = [f'age_{i:.0f}_proportion' for i in range(80, 95, 5)]

df_stats = df_stats.insert_column(-1, pl.Series('age_over80_proportion', df_stats[ages_to_merge].sum_horizontal()))

df_stats = df_stats.drop(ages_to_merge)
```

Drop old proportion age bands to prevent confusion:

```python
df_stats = df_stats.drop(['prop_0-64', 'prop_65-79', 'prop_80+', '0-64', '65-79', '80+'])
```

```python
df_stats.head()
```

Pick out column names for the age proportions:

```python
props_age = [
    'age_less65_proportion', 'age_65_proportion', 'age_70_proportion',
    'age_75_proportion', 'age_over80_proportion'
]
# Set up column names for data we're about to create:
age_numbers = [p.replace('_proportion', '') for p in props_age]
```

Calculate numbers of patients in each age band:

```python
for c, col in enumerate(props_age):
    # Pick out name of new column:
    new_col = age_numbers[c]
    # Create new column:
    # df_stats = df_stats.with_columns((pl.col(col) * pl.col('total_health')).alias(new_col))  # CHANGE from previous version
    df_stats = df_stats.with_columns((pl.col(col) * pl.col('All persons')).alias(new_col))
```

Check results:

```python
df_stats[['All persons', 'total_health'] + props_age + age_numbers].head()
```

## Group by deprivation score

Use quantiles so that each bin contains 10% of the MSOA (to match plot notebook).

The resulting dictionary has keys for which quantile it is and values for the left side (smaller, minimum value in bin) of the IMD bin.

```python
dict_quantiles = {}

for q in np.arange(0.0, 1.01, 0.2):
# for q in np.linspace(0.0, 1.0, 10):
    v = df_stats['IMD2019Score'].quantile(q)
    # Flip the ranking because lower rank (more deprived) should be for
    # higher IMD scores (more deprived).
    q_store = 1.0 - q
    dict_quantiles[round(q_store, 3)] = round(v, 5)
```

```python
dict_quantiles
```

```python
qmin_list = sorted(dict_quantiles.keys())[:-1]

qmin_list
```

Pick out just the values for the left edges of the bins:

```python
imd_bin_left_edges = {}
for k in list(dict_quantiles.keys())[:-1]:
    imd_bin_left_edges[k] = dict_quantiles[k]
```

```python
imd_bin_left_edges
```

Place MSOA into these bins:

```python
# Make columns for the results with a placeholder value:
df_stats = df_stats.with_columns(pl.lit(0.0).alias('depriv_quantile_min'))
df_stats = df_stats.with_columns(pl.lit(0.0).alias('depriv_quantile_max'))

for q, quantile in enumerate(list(dict_quantiles.keys())[:-1]):
    # Pick out the bin edges:
    qmin = list(dict_quantiles.keys())[q+1]
    q0 = dict_quantiles[quantile]
    q1 = dict_quantiles[qmin]
    # Find a mask for the demographic data that contains
    # only MSOA with IMD scores in this bin.
    mask = (df_stats['IMD2019Score'] >= q0) & (df_stats['IMD2019Score'] < q1)
    if q == len(dict_quantiles) - 2:
        # Also allow values at the right edge of the final bin.
        mask = mask | (df_stats['IMD2019Score'] == q1)

    # Update the bin min/max values for these rows:
    df_stats = df_stats.with_columns(
        pl.when((mask))
        .then(qmin)         # replace with bin min
        .otherwise(pl.col('depriv_quantile_min'))  # otherwise keep the existing value
        .name.keep()
    )
    df_stats = df_stats.with_columns(
        pl.when((mask))
        .then(quantile)         # replace with bin min
        .otherwise(pl.col('depriv_quantile_max'))  # otherwise keep the existing value
        .name.keep()
    )
```

Check that data was binned correctly:

```python
df_stats[['IMD2019Score', 'depriv_quantile_min', 'depriv_quantile_max']]
```

## Check for extreme data

View minimum and maximum numbers of people and admissions:

```python
pl.concat((df_stats[['admissions', 'All persons'] + age_numbers].min(),
           df_stats[['admissions', 'All persons'] + age_numbers].max()))
```

View distribution of total number of people in each MSOA:

```python
fig, ax = plt.subplots()
ax.hist(df_stats['All persons'], bins=np.arange(0.0, np.ceil(df_stats['All persons'].max() + 1000), 1000))
ax.set_xlabel('Number of people')
ax.set_ylabel('Number of MSOA')
plt.show()
```

View distribution of admission numbers and pick out the extreme data:

```python
ex_low = df_stats['admissions'].quantile(0.05)
ex_top = df_stats['admissions'].quantile(0.95)

print(f'{ex_low:.1f} {ex_top:.1f}')
```

```python
bins = np.arange(0.0, np.ceil(df_stats['admissions'].max()), 1)

mask = (
    (df_stats['admissions'] <= ex_top) &
    (df_stats['admissions'] >= ex_low) 
)

fig, ax = plt.subplots()
ax.hist(df_stats['admissions'], bins=bins, label='All admissions')
ax.hist(df_stats.filter(mask)['admissions'], bins=bins, histtype='step', edgecolor='k', label='5 to 95% admissions')
ax.set_xlabel('Observed strokes')
ax.set_ylabel('Number of MSOA')
ax.legend()
plt.show()
```

Very few MSOA have unusually high numbers of people, so it's possible that those areas will also have very high admissions numbers.

Most MSOA have between 5 and 21 admissions, but the data outside that range still seem to follow the same distribution as the rest of the data.

So not possible to pick out any anomalous data just by eyeballing the total numbers of people and stroke admissions.
