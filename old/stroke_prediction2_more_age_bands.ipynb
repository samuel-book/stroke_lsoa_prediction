## Set up functions

This multiplies coefficients by the data:

```python
def predict_admissions(x_lists, coeffs):
    """
    x_lists: np.array.
    coeffs: np.array.
    
    Have to have same number of coeffs as x_lists.
    """
    # Predicted admissions:
    yhat = (x_lists * coeffs.reshape(len(coeffs), 1)).sum(axis=0)
    return yhat
    
```

This calculates the sum of the square of the differences between predicted and actual admission numbers:

```python
def find_square_residuals(yhat, y):
    # Difference from actual:
    sqres = (yhat - y)**2.0
    # Sum of differences:
    sum_sqres = sqres.sum()
    return sum_sqres
```

This makes a quick plot of predicted vs admissions numbers:

```python
def plot_y_vs_yhat(y, yhat):
    fig, ax = plt.subplots()

    # Draw the data:
    ax.scatter(y, yhat, s=2, alpha=0.1)

    # Draw a line diagonally:
    ax.set_aspect('equal')
    xmin, xmax = ax.get_xlim()
    ymin, ymax = ax.get_ylim()
    vmin = min([xmin, ymin, 0])
    vmax = max([xmax, ymax, 0])
    ax.plot([vmin, vmax], [vmin, vmax], color='k', linewidth=1)
    ax.set_xlim(min(xmin, 0), xmax)
    ax.set_ylim(min(ymin, 0), ymax)

    ax.set_xlabel('Actual')
    ax.set_ylabel('Predicted')
    
    plt.show()
```

## Optimiser

### Age coefficients

Manually change the coeffs until there's something that resembles a fit to a straight line:

```python
age_props = [
    # 'constant',
    'age_less65_proportion',
    'age_65_proportion',
    'age_70_proportion',
    'age_75_proportion',
    'age_over80_proportion',
]
```

```python
age_coeffs = np.array([-12, 50, 50, 50, 50])
x_lists_age = np.array([df_stats[c] for c in age_props])

yhat = predict_admissions(x_lists_age, age_coeffs)# + df_stats['constant']

plot_y_vs_yhat(df_stats['admissions_plus_constant'], yhat)
```

Run those coeffs through the minimiser:

```python
def main_age(coeffs, args):
    x_lists = args[0]
    admissions = args[1]
    # constant = args[2]#
    yhat = predict_admissions(x_lists, coeffs)# + constant
    sum_sqres = find_square_residuals(yhat, admissions)
    return sum_sqres
```

```python
args = [
    x_lists_age,
    df_stats['admissions_plus_constant'],
    # df_stats['constant'],
]

opt_results = minimize(
    main_age,
    x0=age_coeffs,
    args=args,
    # bounds=[(0.0, None)] * len(age_coeffs),
    method='Nelder-Mead'
)
```

Results:

```python
opt_results
```

```python
age_coeffs_opt = opt_results['x']

age_coeffs_opt
```

Compare with results of the linear regression:

```python
tab = [
    age_props,
    age_coeffs_lr,
    age_coeffs_opt
]

pl.DataFrame(tab, schema=['Variable', 'Linear regression', 'Optimiser'])
```

### IMD coefficients

Fix the age coeffs from above then add in the IMD coeff.

Check whether there's any obvious correlation between the predictions so far and the next data to be added:

```python
def check_corr(df_stats, col, predicted):
    # Calculate line of best fit:
    y_test = (df_stats['admissions_plus_constant']).to_numpy()
    X_test = df_stats[col].to_numpy()
    
    # Add the constant:
    X_test = sm.add_constant(X_test)
    
    model = sm.OLS(y_test, X_test)
    results = model.fit()
    m = results.params[1]
    c = results.params[0]
    r2 = results.rsquared
    line_label = f'y={m:0.4f}x + {c:0.4f}, r^2: {r2:0.4f}'

    
    fig, ax = plt.subplots()
    
    ax.scatter(df_stats[col], yhat, s=2, alpha=0.1)
    
    x_plot = np.array([df_stats[col].min(), df_stats[col].max()])
    y_plot = m * x_plot + c
    ax.plot(x_plot, y_plot, color='k', label=line_label)
    
    ax.set_xlabel(col)
    ax.set_ylabel('Predicted admissions')
    ax.legend()
    
    plt.show()
```

```python
col = 'IMD2019Score'

check_corr(df_stats, col, yhat)
```

Manually change the coeff until there's something that resembles a fit to a straight line:

```python
imd_coeffs = np.array([-0.1])
x_lists_imd = np.array([df_stats['IMD2019Score']])

yhat = predict_admissions(x_lists_age, age_coeffs_opt)
yhatdash = yhat * predict_admissions(x_lists_imd, imd_coeffs)

plot_y_vs_yhat(df_stats['admissions_plus_constant'], yhatdash)
```

Run those coeffs through the minimiser:

```python
def main_imd(coeffs, args):
    x_lists_imd = args[0]
    admissions = args[1]
    yhat = args[2]

    yhatdash = yhat * predict_admissions(x_lists_imd, coeffs)
    
    sum_sqres = find_square_residuals(yhatdash, admissions)
    return sum_sqres
```

```python
yhat = predict_admissions(x_lists_age, age_coeffs_opt)

args = [
    [df_stats['IMD2019Score']],
    df_stats['admissions'],
    yhat,
]

opt_results_imd = minimize(
    main_imd,
    x0=imd_coeffs,
    args=args,
    method='Nelder-Mead',
)#, bounds=[(0.0, None)] * len(imd_coeffs))
```

## Calculate R-squared values

Function to calculate R-squared. This gives the same results as the sklearn built-in.

```python
def calculate_rsquared(y, yhat):
    y_mean = y.mean()
    ss_res = ((yhat - y)**2.0).sum()
    ss_tot = ((y - y_mean)**2.0).sum()
    rsq = 1.0 - ss_res / ss_tot
    return rsq
```

Iterative fits:

```python
rsq_age = calculate_rsquared(df_stats['admissions_plus_constant'], yhat)
rsq_imd_age = calculate_rsquared(df_stats['admissions_plus_constant'], yhatdash)
rsq_health_imd_age = calculate_rsquared(df_stats['admissions_plus_constant'], yhatdashdash)
```
