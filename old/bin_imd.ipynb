# Stroke stats checks stratified by IMD

Check how well various types of data correlate when the full dataset is stratified by Index of Multiple Deprivation.

The correlations plotted from the full dataset can show more scatter than is ideal. By plotting only a subset of MSOA that have an IMD score within a particular ranges, the scatter is reduced and the correlation between the other two variables becomes tighter.

## Code setup

```python
import os
import polars as pl
import numpy as np

from scipy.optimize import minimize
import statsmodels.api as sm  # for linear regression

import matplotlib.pyplot as plt
```

## Load data

Patient demographics by MSOA:

```python
path_to_msoa_stats = os.path.join('data', 'msoa_cleaned.csv')

df_stats = pl.read_csv(path_to_msoa_stats)
```

```python
df_stats.head()
```

Calculate proportion of admissions:

```python
df_stats = df_stats.with_columns((pl.col('admissions') / pl.col('All persons')).alias('prop_admissions'))
```

```python
df_stats.columns
```

Check sum of admissions for Welsh areas:

```python
df_stats.filter(df_stats['country'] == 'W')['admissions'].sum()
```

Welsh data is always zero so remove it.

```python
df_stats = df_stats.filter(df_stats['country'] != 'W')
```

```python
props_health = ['prop_good_health', 'prop_fair health', 'prop_bad health']
props_age = ['age_less65_proportion', 'age_65_proportion', 'age_70_proportion', 'age_75_proportion', 'age_over80_proportion']
```

Recalculate total numbers of patients:

```python
df_stats = df_stats.with_columns((pl.col('good_health') + pl.col('fair health') + pl.col('bad health')).alias('total_health'))
```

Calculate numbers of patients in each age band:

```python
age_numbers = []

for col in props_age:
    new_col = col.replace('_proportion', '')
    age_numbers.append(new_col)
    df_stats = df_stats.with_columns((df_stats[col] * df_stats['total_health']).alias(new_col))
```

```python
df_stats.head()
```

## Decide how to bin IMD

Check the distribution of IMD scores:

```python
fig, ax = plt.subplots()
ax.violinplot(df_stats['IMD2019Score'])
ax.set_ylabel('IMD')
plt.savefig(os.path.join('images', 'violin_imd_england.png'), bbox_inches='tight')
plt.show()
```

Calculate the median and inter-quartile range:

```python
for q in [0.25, 0.50, 0.75]:
    print(f"{q:.2f}: {df_stats['IMD2019Score'].quantile(q):4.1f}")
```

### Even-width bins

First make a histogram with even-width bins:

```python
bins = np.arange(0, max(df_stats['IMD2019Score'])+5, 5)
hist, bins_ = np.histogram(df_stats['IMD2019Score'], bins=bins)

fig, ax = plt.subplots()
ax.stairs(hist, bins_)
ax.set_ylabel('IMD')
plt.savefig(os.path.join('images', 'hist_equalbins_imd_england.png'), bbox_inches='tight')
plt.show()
```

Print the values in the histogram:

```python
for i in range(len(hist)):
    print(bins[i], hist[i])
```

Using groups of IMD with width 5, we find most groups with IMD below around 30 have over 500 patients but that larger IMD groups contain fewer and fewer MSOAs.

By combining multiples groups with larger IMDs, we could make one group with a wider IMD range and a total number of MSOA more comparable with the other groups.

### Quantiles

To ensure that all IMD groups contain the same number of MSOA, use 10 quantiles to bin IMD so each resulting group contains 10% of all MSOAs.

Make the quantiles:

```python
dict_quantiles = {}

for q in np.arange(0.0, 1.01, 0.1):
    v = df_stats['IMD2019Score'].quantile(q)
    dict_quantiles[round(q, 1)] = round(v, 5)
```

```python
dict_quantiles
```

Make the lowest quantiles contain the most deprived areas:

```python
dict_imd = {}

for q in np.arange(0.0, 1.01, 0.1):
    q = np.round(q, 1)
    q1 = np.round(1.0 - q, 1)
    dict_imd[q1] = dict_quantiles[q]
```

## Plot age vs health

Plot age vs health stratified by IMD score:

```python
colour_list = plt.get_cmap('viridis')(np.linspace(0.0, 1.0, len(dict_quantiles)))

fig, ax = plt.subplots()

for q, quantile in enumerate(list(dict_quantiles.keys())[:-1]):
    if q == len(dict_quantiles) - 1:
        q0 = dict_quantiles[quantile]
        mask = (df_stats['IMD2019Score'] >= q0)
        q_str = f'{q0:4.1f} < IMD'
    else:
        q0 = dict_quantiles[quantile]
        q1 = dict_quantiles[list(dict_quantiles.keys())[q+1]]
        mask = (df_stats['IMD2019Score'] >= q0) & (df_stats['IMD2019Score'] < q1)
        q_str = f'{q0:4.1f} <= IMD < {q1:4.1f}'
    df_stats_here = df_stats.filter(mask)
    
    x = 'age_over80_proportion'
    y = 'prop_bad health'
    ax.scatter(df_stats_here[x], df_stats_here[y], label=q_str, s=2, color=colour_list[q])
    ax.set_xlabel(x)
    ax.set_ylabel(y)
    # ax.set_title(q_str)
    ax.legend(bbox_to_anchor=[1.0, 1.0], loc='upper left')
    
plt.savefig(os.path.join('images', 'scatter_age80plus_propbadhealth_by_imd.png'), bbox_inches='tight')
plt.show()
    
```

All combinations of age band and health proportions:

```python
colour_list = plt.get_cmap('viridis')(np.linspace(0.0, 1.0, len(dict_quantiles)))


n_cols = len(props_health)
n_rows = len(props_age)

fig, axs = plt.subplots(n_rows, n_cols, figsize=(14, 14), gridspec_kw={'hspace': 0.3})

for q, quantile in enumerate(list(dict_quantiles.keys())[:-1]):
    if q == len(dict_quantiles) - 1:
        q0 = dict_quantiles[quantile]
        mask = (df_stats['IMD2019Score'] >= q0)
        q_str = f'{q0:4.1f} < IMD'
    else:
        q0 = dict_quantiles[quantile]
        q1 = dict_quantiles[list(dict_quantiles.keys())[q+1]]
        mask = (df_stats['IMD2019Score'] >= q0) & (df_stats['IMD2019Score'] < q1)
        q_str = f'{q0:4.1f} <= IMD < {q1:4.1f}'
    df_stats_here = df_stats.filter(mask)

    # fig.suptitle(q_str)
    
    for iy, y_col in enumerate(props_health):
        for ix, x_col in enumerate(props_age):
            
            # Fit a line to this selection:
            y = (df_stats_here[y_col]).to_numpy()
            X = df_stats_here[x_col].to_numpy()
            # Add the constant:
            X = sm.add_constant(X)
            # Fit line:
            model = sm.OLS(y, X)
            results = model.fit()
            # Pick out coeffs, R^2:
            coeffs = results.params
            r2 = results.rsquared
            m = coeffs[1]
            c = coeffs[0]
            line_label = f'y={m:0.4f}x + {c:0.4f},   r^2: {r2:0.4f}'
        
            xmin = df_stats_here[x_col].min()
            xmax = df_stats_here[x_col].max()
            ymin = m*xmin + c
            ymax = m*xmax + c
            
            col = iy
            row = ix
            ax = axs[row, col]
            # ax.scatter(df_stats[x_col], df_stats[y_col], s=2, color='silver')
            ax.scatter(df_stats_here[x_col], df_stats_here[y_col], s=1, color=colour_list[q], label=q_str)
            # ax.plot([xmin, xmax], [ymin, ymax], color='k', label=line_label)
            ax.set_xlabel(x_col)
            ax.set_ylabel(y_col)
            # ax.legend(bbox_to_anchor=[1.0, 1.0], loc='upper left')

            if row == 0:
                ax.set_xlim(0.45, 1.0)
            else:
                ax.set_xlim(0.0, 0.22)
            if col == 0:
                ax.set_ylim(0.6, 1.1)
                # ax.legend(loc='upper center')
            else:
                ax.set_ylim(0.0, 0.35)
                # ax.legend(loc='upper center')
            if (row == n_rows - 1) & (col == 1):
                ax.legend(loc='upper center', bbox_to_anchor=[0.5, -0.3], ncol=5)
    # axs[0, 1].set_title(q_str, fontsize=18)

    # q0_str = f'{q0:.1f}'.replace('.', 'p')
    
plt.savefig(os.path.join('images', 'scatter_age_health_all_by_imd.png'), bbox_inches='tight')
plt.show()
            
```

```python
colour_list = plt.get_cmap('viridis')(np.linspace(0.0, 1.0, len(dict_quantiles)))

# props_age = ['age_less65_proportion', 'age_65_proportion', 'age_70_proportion', 'age_75_proportion', 'age_over80_proportion']
y_col = 'prop_admissions'

n_cols = len(props_age)
n_rows = int(0.5 * (len(dict_quantiles) - 1))

quantile_names_1 = list(dict_quantiles.keys())[:n_rows]
quantile_names_2 = list(dict_quantiles.keys())[n_rows:-1]

print(quantile_names_1, quantile_names_2)


for w, quantile_names in enumerate([quantile_names_1, quantile_names_2]):
    fig, axs = plt.subplots(n_rows, n_cols, figsize=(14, 20), gridspec_kw={'hspace': 0.6, 'wspace':0.1}, sharey=True)
    for q, quantile in enumerate(quantile_names):
        # if q == len(dict_quantiles) - 1:
        if quantile == list(dict_quantiles.keys())[-1]:
            q0 = dict_quantiles[quantile]
            mask = (df_stats['IMD2019Score'] >= q0)
            q_str = f'{q0:4.1f} < IMD'
        else:
            q0 = dict_quantiles[quantile]
            q1 = dict_quantiles[list(dict_quantiles.keys())[w*n_rows + q+1]]
            mask = (df_stats['IMD2019Score'] >= q0) & (df_stats['IMD2019Score'] < q1)
            q_str = f'{q0:4.1f} <= IMD < {q1:4.1f}'
        df_stats_here = df_stats.filter(mask)
    
        # fig.suptitle(q_str)
        print(q, quantile)
        
        for ix, x_col in enumerate(props_age):
            # Fit a line to this selection:
            y = (df_stats_here[y_col]).to_numpy()
            X = df_stats_here[x_col].to_numpy()
            # Add the constant:
            X = sm.add_constant(X)
            # Fit line:
            model = sm.OLS(y, X)
            results = model.fit()
            # Pick out coeffs, R^2:
            coeffs = results.params
            r2 = results.rsquared
            m = coeffs[1]
            c = coeffs[0]
            line_label = f'y={m:0.4f}x + {c:0.4f},   r^2: {r2:0.4f}'
        
            xmin = df_stats_here[x_col].min()
            xmax = df_stats_here[x_col].max()
            ymin = m*xmin + c
            ymax = m*xmax + c
            
            col = ix
            row = q
            ax = axs[row, col]
            ax.scatter(df_stats[x_col], df_stats[y_col], s=2, color='silver')
            ax.scatter(df_stats_here[x_col], df_stats_here[y_col], s=2, color=colour_list[w*n_rows + q])
            # ax.plot([xmin, xmax], [ymin, ymax], color='k', label=line_label)
            ax.set_xlabel(x_col)
            if col == 0:
                ax.set_ylabel(y_col)
            # ax.legend()
    
            axs[row, 2].set_title(q_str, fontsize=18)
    
    plt.savefig(os.path.join('images', f'scatter_prop_age_admissions_by_imd_{w}.png'), bbox_inches='tight')
    plt.show()
                
```
