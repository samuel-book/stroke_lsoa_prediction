```python
import polars as pl
import os
```

```python
df_msoa = pl.read_csv(os.path.join('data', 'collated_data_regional_msoa11cd.csv'))
```

```python
df_msoa.head()
```

```python
df_admissions = pl.read_csv(os.path.join('data', 'admissions_2017-2019.csv'))

df_admissions.head()
```

```python
df_lsoa_msoa = pl.read_csv(os.path.join('data', 'lsoa_to_msoa.csv'))

df_lsoa_msoa.head()
```

```python
df_admissions = df_admissions.join(df_lsoa_msoa[['lsoa11nm', 'msoa11cd', 'msoa11nm']], left_on='area', right_on='lsoa11nm', how='left')
```

```python
df_admissions.head()
```

```python
df_msoa_admissions = df_admissions.group_by('msoa11cd').sum()[['msoa11cd', 'Admissions']]
```

```python
df_msoa_admissions.head()
```

```python
df_msoa = df_msoa.join(df_msoa_admissions, on='msoa11cd')
```

```python
df_msoa = df_msoa.join(df_admissions[['msoa11cd', 'msoa11nm']], on='msoa11cd')
```

```python
df_msoa.head()
```

```python
df_msoa_admissions.sum()
```
